# 2.naarray的属性

```python
arr = np.array(1)
print(arr)
print('shape:', arr.shape)
print('dimension:', arr.ndim)
print('size:', arr.size)
print('type:', arr.dtype)
```

```python
arr = np.array([[1.09, 2, 3], [1, 2, 3]])
print(arr)
print('shape:', arr.shape)
print('dimension:', arr.ndim)
print('size:', arr.size)
print('type:', arr.dtype)
```

## 2.2ndarray的创建

```python
arr = np.array([[1.09, 2, 3], [1, 2, 3]])
print(arr)
print('dimension:', arr.ndim)
```

```python
arr1 = np.copy(arr)
print(arr1)
arr1[0, 1] = 8
print(arr1)
```

```python
arr = np.zeros((0), dtype=int)
print(arr)
```

```python
arr = np.ones((2, 3), dtype=int)
print(arr)
```
